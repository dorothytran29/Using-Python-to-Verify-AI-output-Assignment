```python
# Read the dataset and store it as data2026
data2026=pd.read_csv('2026PartYear.csv', encoding='utf-8')

# Read the dataset and store it as data2025
data2025=pd.read_csv('2025.csv', encoding='utf-8')
new=pd.concat([data2026,data2025])
```

```python
# Filter for Total U.S.
us_2025 = data2025[data2025['Geography'] == 'Total U.S.']
us_2026 = data2026[data2026['Geography'] == 'Total U.S.']
```

```python
# Verify total U.S. volume increase from 2025 to 2026
volume2025 = us_2025['Total Bulk and Bags'].sum()
volume2026 = us_2026['Total Bulk and Bags'].sum()
# Calculate volume increase
volume_increase = volume2026 - volume2025
# Calculate percent increase
percent_increase = (volume_increase / volume2025) * 100

print("2025 Total U.S. Volume:", volume2025)
print("2026 Total U.S. Volume:", volume2026)
print("Volume Increase:", volume_increase)
print("Percent Increase:", percent_increase, "%")
```

```python
# Verify Volume-Weighted ASP 
# Formula: sum(ASP * Volume) / sum(Volume)
asp_2025 = ((us2025['ASP Current Year'] * us2025['Total Bulk and Bags']).sum() / us2025['Total Bulk and Bags'].sum())
asp_2026 = ((us2026['ASP Current Year'] * us2026['Total Bulk and Bags']).sum() / us2026['Total Bulk and Bags'].sum())
# Dollar change
asp_change = asp_2026 - asp_2025
# Percent change
asp_percent_change = ((asp_2026 - asp_2025) / asp_2025) * 100

print("2025 ASP:", round(asp_2025, 2))
print("2026 ASP:", round(asp_2026, 2))
print("ASP Percent Change:", round(asp_percent_change, 2), "%")
```

```python
# Verify Bagged Volume Increase
bagged_2025 = 782  # in millions
bagged_2026 = 920  # in millions
bagged_increase_pct = ((bagged_2026 - bagged_2025) / bagged_2025) * 100
print("Bagged Volume Increase:", round(bagged_increase_pct, 2), "%")
```

```python
# Filter both years to the same time period: Periods 1-32
data2025_32 = data2025[data2025['Period'] <= 32]
data2026_32 = data2026[data2026['Period'] <= 32]
```

```python
# Verify PLU Volume Changes
for plu in ['4046', '4225', '4770']:
    plu_2025 = data2025_32[plu].sum()
    plu_2026 = data2026_32[plu].sum()
    percent_change = ((plu_2026 - plu_2025) / plu_2025) * 100
    print(f"PLU {plu}")
    print(f"2025 Volume: {plu_2025:,.2f}")
    print(f"2026 Volume: {plu_2026:,.2f}")
    print(f"Volume Change: {percent_change:.2f}%")
    print()
```

```python
# To verify each PLU's share
total_plu_2025 = (data2025_32['4046'].sum() + data2025_32['4225'].sum() + data2025_32['4770'].sum())
total_plu_2026 = (data2026_32['4046'].sum() + data2026_32['4225'].sum() + data2026_32['4770'].sum())

for plu in ['4046', '4225', '4770']:

    share_2025 = (data2025_32[plu].sum() / total_plu_2025) * 100
    share_2026 = (data2026_32[plu].sum() / total_plu_2026) * 100

    print(f"PLU {plu}")
    print(f"2025 Share: {share_2025:.2f}%")
    print(f"2026 Share: {share_2026:.2f}%")
    print()
```

```python
# Verify Conventional Volume
# Conventional volume
conv_2025 = data2025_32[data2025_32['Type'] == 'Conventional']['Total Bulk and Bags'].sum()
conv_2026 = data2026_32[data2026_32['Type'] == 'Conventional']['Total Bulk and Bags'].sum()
conv_change = ((conv_2026 - conv_2025) / conv_2025) * 100
print("Conventional Volume Change:", round(conv_change, 2), "%")

# Organic volume
org_2025 = data2025_32[data2025_32['Type'] == 'Organic']['Total Bulk and Bags'].sum()
org_2026 = data2026_32[data2026_32['Type'] == 'Organic']['Total Bulk and Bags'].sum()
org_change = ((org_2026 - org_2025) / org_2025) * 100
print("Organic Volume Change:", round(org_change, 2), "%")
```

```python
# Verify Regional Volume Growth
region_2025 = data2025_32.groupby('Geography')['Total Bulk and Bags'].sum()
region_2026 = data2026_32.groupby('Geography')['Total Bulk and Bags'].sum()

regional_volume = pd.DataFrame({'2025 Volume': region_2025,'2026 Volume': region_2026})
regional_volume['Volume Change %'] = ((regional_volume['2026 Volume'] - regional_volume['2025 Volume']) / regional_volume['2025 Volume']) * 100
display(regional_volume.sort_values('2026 Volume', ascending=False))
```
